# RL Simulator
This notebook documents the approach to building the autoscaler simulator and verifies that it is functioning by taking random actions in the environment.

* The environment replays real Google workload data one minute at a time, using CPU usage as the primary driver of demand.

* At each step, the agent observes the current CPU load and the number of machines currently running.

* The agent chooses one of three actions: scale down, hold, or scale up, subject to capacity limits and a cooldown period that prevents rapid scaling decisions.

* Raw CPU usage from the dataset is too small to represent realistic cluster load, so the environment applies a demand scaling factor to create meaningful, cluster-level workload pressure.

* Utilization is derived from the scaled demand and current capacity, determining how “busy” the cluster is relative to its size.

* The reward function balances three objectives: minimizing cost, avoiding SLA violations from excessive utilization, and keeping utilization close to a target operating point.

* The environment progresses sequentially through the workload dataset, allowing the agent to learn scaling policies that react to realistic changes in demand.

* An episode ends when the workload data is fully consumed, simulating a complete run through a production-like time series.

## Run Simulation Taking Random Actions

```python
# Define the number of simulation steps
num_simulation_steps = 100

# Initialize the environment
env = AutoScalingEnv(df_usage)

# Reset the environment to get the initial observation
obs, info = env.reset()

# Lists to store results
observations = [obs]
rewards = []
actions = []
infos = [info]

# Start the simulation loop
for step in range(num_simulation_steps):

    # Random action (for baseline)
    action = env.action_space.sample()

    # Take environment step
    next_obs, reward, terminated, truncated, info = env.step(action)

    # Store results
    observations.append(next_obs)
    rewards.append(reward)
    actions.append(action)
    infos.append(info)

    # Check termination
    if terminated or truncated:
        print(f"Episode finished after {step+1} steps.")
        break

    obs = next_obs

print("Simulation complete.")
```

```python
import numpy as np

# Create a DataFrame from the collected data
simulation_results_df = pd.DataFrame({
    'step': range(num_simulation_steps),
    'action': actions,
    'reward': rewards
})

# Extract information from the 'infos' list and add as columns
simulation_results_df['current_capacity'] = [info['current_capacity'] for info in infos[1:]]
simulation_results_df['utilization'] = [info['utilization'] for info in infos[1:]]
simulation_results_df['estimated_total_cpu_load'] = [info['demand_cpu'] for info in infos[1:]]
# Extract reward components if available, handling potential missing keys
simulation_results_df['cost_penalty'] = [info['reward_components'].get('cost_penalty', np.nan) for info in infos[1:]]
simulation_results_df['sla_penalty'] = [info['reward_components'].get('sla_penalty', np.nan) for info in infos[1:]]
simulation_results_df['util_deviation_penalty'] = [info['reward_components'].get('util_deviation_penalty', np.nan) for info in infos[1:]]


# Set the step number as the index
simulation_results_df.set_index('step', inplace=True)

# Display the first few rows of the DataFrame
display(simulation_results_df.head())
```

## Visualize results



```python
import matplotlib.pyplot as plt

# Create a figure and a set of subplots
fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(12, 15), sharex=True)

# Plot Current Capacity and Estimated Total CPU Load
axes[0].plot(simulation_results_df.index, simulation_results_df['current_capacity'], label='Current Capacity')
axes[0].plot(simulation_results_df.index, simulation_results_df['estimated_total_cpu_load'], label='Estimated Total CPU Load')
axes[0].set_ylabel('Capacity / Load')
axes[0].set_title('Autoscaling Simulation Results')
axes[0].legend()
axes[0].grid(True)

# Plot Utilization and Target Utilization
axes[1].plot(simulation_results_df.index, simulation_results_df['utilization'], label='Utilization')
axes[1].axhline(y=env.target_utilization, color='g', linestyle='--', label='Target Utilization')
axes[1].axhline(y=env.sla_threshold, color='r', linestyle='--', label='SLA Threshold')
axes[1].set_ylabel('Utilization')
axes[1].legend()
axes[1].grid(True)

# Plot Reward
axes[2].plot(simulation_results_df.index, simulation_results_df['reward'], label='Reward')
axes[2].set_xlabel('Step')
axes[2].set_ylabel('Reward')
axes[2].legend()
axes[2].grid(True)

# Adjust layout to prevent overlapping
plt.tight_layout()

# Display the plots
plt.show()
```
